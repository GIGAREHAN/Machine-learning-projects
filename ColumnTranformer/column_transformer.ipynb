# COVID-19 Toy Dataset Preprocessing

This notebook demonstrates how to build an end-to-end data preprocessing pipeline using scikit-learn's `ColumnTransformer` on a toy COVID-19 dataset.

### Objectives:
- **Handle Missing Values**: Impute missing numerical measurements (e.g., body temperature/fever).
- **Ordinal Encoding**: Correctly map ordered categorical features (such as cough severity: 'Mild' vs. 'Strong').
- **One-Hot Encoding**: Encode nominal categorical features (e.g., gender and city) to prevent introducing artificial order.
- **Keep Remaining Features**: Ensure numeric predictors like `age` pass through the pipeline untouched.

### Pipeline Overview:
1. **Data Loading & Splitting**: Segment the dataset into training and testing sets.
2. **Feature Transformation**: Leverage `ColumnTransformer` to target specific columns with specialized estimators.
3. **Validation**: Test the dimensions of the final transformed feature matrices.

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['has_covid']), df['has_covid'], test_size=0.20)
```

```python
X_train
```

```python
#Using column transformer

from sklearn.compose import ColumnTransformer
```

```python
from pandas.core.arrays import sparse
transformer = ColumnTransformer(transformers = [
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city'])
], remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
